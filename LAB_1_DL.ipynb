**Activity 2 – Creating Tensors**

```python
import torch
A = torch.tensor([[1,2],
                  [3,4]])

print(A)
print(type(A))
print(A.shape)
print(A.dtype)
```

```python
import tensorflow as tf

A = tf.constant([[1,2],
                 [3,4]])

print(A)
print(type(A))
print(A.shape)
print(A.dtype)
```

**Activity 3 – Tensor Operations**

```python
import torch

A = torch.tensor([[1,2],
                  [3,4]])

B = torch.tensor([[5,6],
                  [7,8]])

print("Addition")
print(A+B)

print("Subtraction")
print(A-B)

print("Multiplication")
print(A*B)

print("Matrix Multiplication")
print(torch.matmul(A,B))
```

```python
import tensorflow as tf

A = tf.constant([[1,2],
                 [3,4]])

B = tf.constant([[5,6],
                 [7,8]])

print(A+B)
print(A-B)
print(A*B)
print(tf.matmul(A,B))
```

**Activity 4 – Activation Functions**

```python
import torch
import matplotlib.pyplot as plt

x = torch.linspace(-5,5,200)

relu = torch.relu(x)
sigmoid = torch.sigmoid(x)
tanh = torch.tanh(x)

plt.figure(figsize=(8,5))

plt.plot(x, relu, label="ReLU")
plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, tanh, label="Tanh")

plt.legend()
plt.grid(True)

plt.show()
```

**Activity 5 – Build a Perceptron (PyTorch)**

```python
import torch
import torch.nn as nn

model = nn.Sequential(

    nn.Linear(2,1),

    nn.Sigmoid()

)

print(model)
```

**Activity 6 – Build a Perceptron (TensorFlow)**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([

Dense(1,
      activation="sigmoid",
      input_shape=(2,))

])

model.summary()
```

**Activity 7 – Forward Propagation**

```python
import torch

model = nn.Sequential(

nn.Linear(2,1),

nn.Sigmoid()

)

x = torch.tensor([[1.0,2.0]])

output = model(x)

print(output)
```

```python
import tensorflow as tf

model = Sequential([

Dense(1,
activation='sigmoid',
input_shape=(2,))

])

x = tf.constant([[1.0,2.0]])

prediction = model(x)

print(prediction)
```

**Activity 8 – Build an MLP**

```python
import torch.nn as nn

model = nn.Sequential(

nn.Linear(4,8),

nn.ReLU(),

nn.Linear(8,4),

nn.ReLU(),

nn.Linear(4,1),

nn.Sigmoid()

)

print(model)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([

Dense(8,
activation='relu',
input_shape=(4,)),

Dense(4,
activation='relu'),

Dense(1,
activation='sigmoid')

])

model.summary()
```

**Activity 10 – Prediction on Dummy Data**

```python
import numpy as np

X = np.random.rand(5,4)

print(X)
```

```python
prediction = model.predict(X)

print(prediction)
```

```python
import torch

X = torch.tensor(X,
dtype=torch.float32)

prediction = model(X)

print(prediction)
```
